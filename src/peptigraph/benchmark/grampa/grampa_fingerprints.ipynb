```python
import pandas as pd

from peptigraph.main import run_experiment
```

```python
def run_single_peptide_experiment(dataset_name, loader):
    base_params = {
        'model_type': 'rf',
        'use_pipeline': True
    }

    fingerprint_configs = [
        # Single fingerprints
        ['ecfp'],
        ['maccs'],
        ['tt'],
        ['ap'],
        ['layered'],
        
        # Two fingerprint combinations
        ['ecfp', 'tt'],
        ['ecfp', 'maccs'],
        ['ecfp', 'ap'],
        ['ecfp', 'layered'],
        ['ap', 'tt'],
        ['tt', 'layered'],
        
        # Three fingerprint combinations
        ['ecfp', 'tt', 'ap'],
        ['ecfp', 'maccs', 'layered'],
        ['ecfp', 'tt', 'layered'],
        ['tt', 'ap', 'layered'],
        
        # Four fingerprint combinations
        ['ecfp', 'tt', 'ap', 'layered'],
        ['ecfp', 'maccs', 'ap', 'layered']
    ]
    
    print(f"\nProcessing {dataset_name}")
    print("=" * 50)
    
    dataset_results = {}
    
    for fp_types in fingerprint_configs:
        fp_display = ' + '.join(fp.upper() for fp in fp_types)
        fp_key = tuple(fp_types)
        
        print(f"\nTesting {fp_display} fingerprint")
        params = {**base_params, 'fingerprint_types': fp_types}
        
        try:
            results = run_experiment(loader, params)
            dataset_results[fp_key] = results
        except Exception as e:
            print(f"Error with {fp_display}: {str(e)}")
            dataset_results[fp_key] = None
    
    print(f"\nSummary for {dataset_name}:")
    print("-" * 30)
    for fp_types, metrics in dataset_results.items():
        if metrics:
            fp_display = ' + '.join(fp.upper() for fp in fp_types)
            print(f"\n{fp_display}:")
            for metric, score in metrics.items():
                print(f"{metric.upper()}: {score:.2%}")
                
    return dataset_results

def load_grampa():  
    df = pd.read_csv('grampa_reduced.csv')
    return df['smiles'].values, df['target'].values

def analyze_grampa():
    results = run_single_peptide_experiment('Grampa', load_grampa)
    return results
```

```python
results = analyze_grampa()
```

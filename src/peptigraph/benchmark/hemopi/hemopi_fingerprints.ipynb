```python
import pandas as pd

from peptigraph.benchmark.main import run_experiment
```

```python
def run_single_hemopi_experiment(dataset_name, loader):
    base_params = {
        'model_type': 'rf',
        'use_pipeline': True
    }

    # Original working fingerprint configurations
    fingerprint_configs = [
        # Single fingerprints
        ['ecfp'],
        ['maccs'],
        ['tt'],
        ['ap'],
        ['layered'],
        
        # Two fingerprint combinations
        ['ecfp', 'tt'],
        ['ecfp', 'maccs'],
        ['ecfp', 'ap'],
        ['ecfp', 'layered'],
        ['ap', 'tt'],
        ['tt', 'layered'],
        
        # Three fingerprint combinations
        ['ecfp', 'tt', 'ap'],
        ['ecfp', 'maccs', 'layered'],
        ['ecfp', 'tt', 'layered'],
        ['tt', 'ap', 'layered'],
        
        # Four fingerprint combinations
        ['ecfp', 'tt', 'ap', 'layered'],
        ['ecfp', 'maccs', 'ap', 'layered']
    ]
    
    print(f"\nProcessing {dataset_name}")
    print("=" * 50)
    
    dataset_results = {}
    
    for fp_types in fingerprint_configs:
        fp_display = ' + '.join(fp.upper() for fp in fp_types)
        fp_key = tuple(fp_types)
        
        print(f"\nTesting {fp_display} fingerprint")
        params = {**base_params, 'fingerprint_types': fp_types}
        
        try:
            results = run_experiment(loader, params)
            dataset_results[fp_key] = results
        except Exception as e:
            print(f"Error with {fp_display}: {str(e)}")
            dataset_results[fp_key] = None
    
    # Print summary for this dataset
    print(f"\nSummary for {dataset_name}:")
    print("-" * 30)
    for fp_types, metrics in dataset_results.items():
        if metrics:
            fp_display = ' + '.join(fp.upper() for fp in fp_types)
            print(f"\n{fp_display}:")
            for metric, score in metrics.items():
                print(f"{metric.upper()}: {score:.2%}")
                
    return dataset_results
```

```python
def load_hemopi_1():
    df = pd.read_csv('C:\Python\semestr_1\PeptiGraph\data\smiles\hemopi_1.csv')
    return df['smiles'].values, df['target'].values

def load_hemopi_2():
    df = pd.read_csv('C:\Python\semestr_1\PeptiGraph\data\smiles\hemopi_2.csv')
    return df['smiles'].values, df['target'].values

def load_hemopi_3():
    df = pd.read_csv('C:\Python\semestr_1\PeptiGraph\data\smiles\hemopi_3.csv')
    return df['smiles'].values, df['target'].values
```

```python
def analyze_hemopi_1():
    results_hemopi_1 = run_single_hemopi_experiment('HemoPI-1', load_hemopi_1)
    return results_hemopi_1

def analyze_hemopi_2():
    results_hemopi_2 = run_single_hemopi_experiment('HemoPI-2', load_hemopi_2)
    return results_hemopi_2

def analyze_hemopi_3():
    results_hemopi_3 = run_single_hemopi_experiment('HemoPI-3', load_hemopi_3)
    return results_hemopi_3
```
